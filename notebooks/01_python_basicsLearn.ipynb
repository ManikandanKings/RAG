# Python Data Structures Comparison

| Feature            | List      | Tuple     | Set       | Dictionary          |
|--------------------|-----------|-----------|-----------|---------------------|
| Ordered            | ✅ Yes    | ✅ Yes    | ❌ No     | ✅ Yes (Python 3.7+) |
| Mutable            | ✅ Yes    | ❌ No     | ✅ Yes    | ✅ Yes              |
| Allows Duplicates  | ✅ Yes    | ✅ Yes    | ❌ No     | Keys ❌, Values ✅   |
| Indexed            | ✅ Yes    | ✅ Yes    | ❌ No     | Access by Key       |
| Syntax             | `[]`      | `()`      | `{}`      | `{key:value}`       |
| Fast Lookup        | ❌ O(n)   | ❌ O(n)   | ✅ O(1)   | ✅ O(1)

##Functions and Parameters

```python
def greet(name, greeting='Hello'):
    return f'{greeting}, {name}'

print(greet('Abdul'))


def add(a: int, b: int) -> int:
    return a + b

result = add(10, 20)
print(result)
```

##Classes and Objects

```python
class Employee:
  
    def __init__(self, id, name):
        self.id = id
        self.name = name

    def display(self):
        print(f"Id: {self.id}")
        print(f"Name: {self.name}")

employee = Employee(1, "Abdul")

employee.display()
```

# Python Access Modifiers

| Modifier  | Syntax   | Same Class | Child Class | Outside Class |
|-----------|----------|------------|-------------|---------------|
| Public    | `name`   | ✅          | ✅           | ✅             |
| Protected | `_name`  | ✅          | ✅           | ⚠️ Convention only |
| Private   | `__name` | ✅          | ❌ Directly  | ❌ Directly    |


##Exceptions

```python
try:
    result=10/0
except ZeroDivisionError as ex:
    print('Error:', ex)
except Exception as ex:
    print('Error:', ex)
finally:
    print('Done')
```
