```python
from sentence_transformers import SentenceTransformer, util

# Load model
model = SentenceTransformer("all-MiniLM-L6-v2")
model




```

```python
# Sentences and query
sentences = [
    "I want to buy a car",
    "I am looking to purchase an automobile",
    "The weather is hot today",
    "sachin tendulkar is a great cricketer"
]
query = "i want to buy a car"

# Embed sentences and query
# sentence_embeddings = model.encode(sentences, convert_to_tensor=True)
# query_embedding = model.encode(query, convert_to_tensor=True)
sentence_embeddings = model.encode(sentences)
# print(sentence_embeddings[0])
query_embedding = model.encode(query)


# Calculate cosine similarity
cosine_scores = util.cos_sim(query_embedding, sentence_embeddings)

# Print similarity scores
print("\nSimilarity scores:")
for i, sentence in enumerate(sentences):
    print(f"Query vs '{sentence}': {cosine_scores[0][i]:.4f}")

# Find most similar sentence
best_idx = cosine_scores.argmax()
print("\nMost similar sentence:", sentences[best_idx])
```
